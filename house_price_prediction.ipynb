# House Price Prediction System
An end-to-end regression project using Pandas, NumPy, Scikit-learn, and Linear Regression.

**Dataset note:** The included data is synthetic demo data for learning. Replace it with a properly sourced housing dataset before drawing real-market conclusions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path('data/house_prices.csv')
df = pd.read_csv(DATA_PATH)
print('Dataset shape:', df.shape)
display(df.head())
display(df.describe(include='all'))
```

## 1. Explore and clean the data

```python
print(df.isna().sum())
df = df.drop_duplicates().dropna().copy()
df['location'] = df['location'].astype(str).str.strip()
for col in ['area_sqft','bedrooms','bathrooms','age_years','price_inr']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df = df.dropna()
df = df[(df['area_sqft'] > 0) & (df['price_inr'] > 0)]
print('Cleaned shape:', df.shape)
display(df.head())
```

## 2. Visualize relationships

```python
plt.figure(figsize=(8,5))
plt.scatter(df['area_sqft'], df['price_inr'])
plt.xlabel('Area (sq ft)'); plt.ylabel('Price (INR)')
plt.title('House Area vs Price'); plt.tight_layout(); plt.show()

df.groupby('location')['price_inr'].mean().sort_values().plot(kind='bar')
plt.ylabel('Average price (INR)'); plt.title('Average Price by Location')
plt.tight_layout(); plt.show()
```

## 3. Train and evaluate Linear Regression
The reusable training script applies one-hot encoding to the categorical location field and fits preprocessing only on the training data.

```python
from train_model import train_and_save_model
model, metrics = train_and_save_model(data_path=DATA_PATH)
print(metrics)
```

## 4. Make a sample prediction

```python
sample = pd.DataFrame([{
    'area_sqft': 1200,
    'bedrooms': 2,
    'bathrooms': 2,
    'age_years': 5,
    'location': 'Hyderabad'
}])
prediction = model.predict(sample)[0]
print(f'Estimated house price: ₹{prediction:,.0f}')
```
